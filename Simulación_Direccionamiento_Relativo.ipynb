```python
# @title
import ipywidgets as widgets
from IPython.display import display, HTML

# -------------------------
# ENTRADAS
# -------------------------

base = widgets.IntText(
    value=1000,
    description='Base:',
    style={'description_width': '80px'}
)

offset = widgets.IntText(
    value=256,
    description='Offset:',
    style={'description_width': '80px'}
)

boton = widgets.Button(
    description='▶ CALCULAR Y ACCEDER A RAM',
    button_style='info',
    layout=widgets.Layout(width='280px', height='45px')
)

salida = widgets.Output()


# -------------------------
# SIMULACIÓN
# -------------------------

def simular(b):

    direccion = base.value + offset.value

    # Posiciones alrededor de la dirección encontrada
    posiciones = [
        direccion - 8,
        direccion - 4,
        direccion,
        direccion + 4,
        direccion + 8
    ]

    bloques_ram = ""

    for pos in posiciones:

        if pos == direccion:

            bloques_ram += f"""
            <div style="
                background:#00cfff;
                color:#00172b;
                padding:14px;
                margin:7px;
                border-radius:8px;
                font-weight:bold;
                box-shadow:0 0 15px #00cfff;
            ">
                ▶ {pos} &nbsp;&nbsp; DATO ENCONTRADO
            </div>
            """

        else:

            bloques_ram += f"""
            <div style="
                background:#142a46;
                color:white;
                padding:14px;
                margin:7px;
                border-radius:8px;
                border:1px solid #34506f;
            ">
                {pos}
            </div>
            """

    html = f"""

    <div style="
        background:#07182c;
        color:white;
        padding:25px;
        border-radius:15px;
        font-family:Arial;
        width:850px;
    ">

        <h2 style="color:#21d4ff;">
        Simulación de Direccionamiento Relativo
        </h2>

        <div style="
            display:flex;
            gap:30px;
            margin-top:20px;
        ">

            <!-- CPU -->

            <div style="
                background:#10243d;
                padding:20px;
                border-radius:12px;
                width:45%;
            ">

                <h3 style="color:#21d4ff;">
                ⚙ CPU
                </h3>

                <p>Proceso en ejecución</p>

                <div style="
                    background:#07182c;
                    padding:12px;
                    border-radius:8px;
                    margin-top:15px;
                ">

                    Dirección Base

                    <h2 style="color:#21d4ff;">
                    {base.value}
                    </h2>

                </div>

                <div style="
                    background:#07182c;
                    padding:12px;
                    border-radius:8px;
                    margin-top:10px;
                ">

                    Desplazamiento (Offset)

                    <h2 style="color:#21d4ff;">
                    + {offset.value}
                    </h2>

                </div>

                <h3 style="
                    margin-top:20px;
                    color:#00ffae;
                ">

                {base.value} + {offset.value}
                = {direccion}

                </h3>

                <p>
                Dirección efectiva calculada
                </p>

            </div>


            <!-- RAM -->

            <div style="
                background:#10243d;
                padding:20px;
                border-radius:12px;
                width:45%;
            ">

                <h3 style="color:#21d4ff;">
                ▣ MEMORIA RAM
                </h3>

                {bloques_ram}

            </div>

        </div>


        <div style="
            background:#0d3551;
            margin-top:20px;
            padding:15px;
            border-radius:10px;
            text-align:center;
        ">

            CPU
            &nbsp; ➜ &nbsp;

            <b style="color:#21d4ff;">
            {base.value} + {offset.value}
            </b>

            &nbsp; ➜ &nbsp;

            RAM

            &nbsp; ➜ &nbsp;

            <b style="color:#00ffae;">
            Dirección {direccion}
            </b>

        </div>

    </div>

    """

    with salida:

        salida.clear_output()

        display(HTML(html))


boton.on_click(simular)


# -------------------------
# INTERFAZ
# -------------------------

display(
    HTML("""
    <h2>🧠 Simulador de Memoria</h2>
    <p>
    Ingresa la dirección base y el desplazamiento:
    </p>
    """)
)

display(base)
display(offset)
display(boton)
display(salida)
```
